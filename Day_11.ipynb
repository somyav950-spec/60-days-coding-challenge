```python
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

print("Libraries imported successfully!")
```

```python
documents = [
    "Data analytics is the process of examining data to find useful patterns and insights.",
    "Python is widely used in data analytics because it provides powerful libraries for data processing.",
    "Pandas is a Python library used for data manipulation and analysis.",
    "NumPy provides support for numerical operations and array processing in Python.",
    "Excel is commonly used for organizing, analyzing and visualizing business data.",
    "Power BI is a business intelligence tool used to create interactive dashboards and reports.",
    "SQL is used to retrieve, filter and analyze data stored in relational databases.",
    "Data visualization helps users understand complex information through charts and graphs.",
    "A dashboard presents important metrics and insights in an easy to understand format.",
    "Machine learning allows computers to learn patterns from data and make predictions.",
    "Supervised learning uses labeled data to train machine learning models.",
    "Unsupervised learning finds hidden patterns in data without labeled outputs.",
    "Artificial intelligence refers to computer systems that perform tasks requiring human-like intelligence.",
    "Natural language processing enables computers to process and understand human language.",
    "Data cleaning involves handling missing values, duplicates and incorrect data.",
    "Missing values can be handled using methods such as deletion or statistical imputation.",
    "Exploratory data analysis is used to understand the structure and characteristics of a dataset.",
    "Statistics helps data analysts summarize data and identify relationships between variables.",
    "A data analyst uses tools such as Python, SQL, Excel and Power BI to analyze information.",
    "Data science combines statistics, programming and machine learning to extract knowledge from data."
]

print("Number of documents:", len(documents))
```

```python
vectorizer = TfidfVectorizer()

corpus_matrix = vectorizer.fit_transform(documents)

print("TF-IDF matrix shape:", corpus_matrix.shape)
```

```python
def retrieve(query, corpus_matrix, top_k=3):
    query_vector = vectorizer.transform([query])

    similarity_scores = cosine_similarity(
        query_vector,
        corpus_matrix
    )[0]

    max_score = np.max(similarity_scores)

    if max_score < 0.1:
        return "No relevant document found"

    top_indices = similarity_scores.argsort()[-top_k:][::-1]

    results = []

    for index in top_indices:
        results.append({
            "document": documents[index],
            "score": round(similarity_scores[index], 3)
        })

    return results
```

```python
query = "What is Python used for in data analysis?"

results = retrieve(query, corpus_matrix)

print(results)
```

```python
test_queries = [
    "Python for data analysis",
    "working with tables and data",
    "creating business dashboards",
    "querying databases",
    "handling missing data",
    "finding patterns in data",
    "learning from examples",
    "data charts",
    "how to cook pasta",
    "weather forecast for tomorrow"
]

for i, query in enumerate(test_queries, 1):
    print(f"\nQuery {i}: {query}")
    print("-" * 60)

    result = retrieve(query, corpus_matrix, top_k=3)

    if isinstance(result, str):
        print(result)
    else:
        for rank, item in enumerate(result, 1):
            print(f"{rank}. Score: {item['score']}")
            print(f"   {item['document']}")
```

```python
failure_analysis = {
    "working with tables and data":
        "The retrieval may prefer Python or general data documents because the query uses broad vocabulary.",

    "finding patterns in data":
        "The query is broad and may match several documents because words like data and patterns occur across the knowledge base.",

    "learning from examples":
        "The query is ambiguous because it does not explicitly mention machine learning or supervised learning.",

    "data charts":
        "The short query may produce weaker matching because the exact vocabulary differs from the document wording."
}

for query, diagnosis in failure_analysis.items():
    print("Query:", query)
    print("Diagnosis:", diagnosis)
    print()
```

```python
out_of_domain_queries = [
    "how to cook pasta",
    "weather forecast for tomorrow"
]

for query in out_of_domain_queries:
    print("\nQuery:", query)
    print("Result:", retrieve(query, corpus_matrix))
```

```python
threshold_test = [
    "how to cook pasta",
    "weather forecast for tomorrow",
    "quantum physics experiment"
]

for query in threshold_test:
    query_vector = vectorizer.transform([query])

    scores = cosine_similarity(
        query_vector,
        corpus_matrix
    )[0]

    print("\nQuery:", query)
    print("Highest similarity score:", round(np.max(scores), 3))
    print("Retrieved:", retrieve(query, corpus_matrix))
```

TF-IDF retrieval mainly depends on the words shared between the query and documents.
If a user uses a synonym that does not appear in the knowledge base, the system may fail to retrieve the correct document.
For example, a document may use the word "data analysis" while a user searches for "examining information".
Because these words are different, TF-IDF may assign a low similarity score even though the meanings are similar.
This limitation shows why semantic embeddings are useful because embeddings can capture meaning and relationships between words.

## Conclusion

The retrieval engine was built using TF-IDF and cosine similarity.
A knowledge base of 20 documents was created and vectorised.
The retrieve() function returns the top 3 relevant documents with similarity scores.
The system was tested using 10 different queries, including ambiguous and out-of-domain queries.
A relevance threshold of 0.1 was implemented to reject very low-quality matches.
The experiment also showed that vocabulary mismatch can cause TF-IDF retrieval failures.
Semantic embeddings can improve retrieval by understanding meaning rather than relying only on exact word overlap.
