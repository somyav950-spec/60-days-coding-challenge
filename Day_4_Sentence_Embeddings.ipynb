```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
sentences = [
    "I love learning artificial intelligence.",
    "I enjoy studying machine learning.",
    "Python is my favorite programming language.",
    "The weather is very pleasant today.",
    "I like eating pizza with my friends."
]

print(sentences)
```

```python
embeddings = model.encode(sentences)

print("Number of sentences:", len(sentences))
print("Embedding shape:", embeddings.shape)
```

```python
similarity_matrix = cosine_similarity(embeddings)

print(similarity_matrix)
```

```python
for i in range(len(sentences)):
    for j in range(i + 1, len(sentences)):
        score = similarity_matrix[i][j]
        print(f"Sentence {i+1} & Sentence {j+1}: {score:.4f}")
```

```python
max_score = -1
best_pair = None

for i in range(len(sentences)):
    for j in range(i + 1, len(sentences)):
        score = similarity_matrix[i][j]

        if score > max_score:
            max_score = score
            best_pair = (i, j)

print("Most similar sentences:")
print(sentences[best_pair[0]])
print(sentences[best_pair[1]])
print("Similarity Score:", round(max_score, 4))
```

Reflection: I learned that sentence embeddings capture the meaning of sentences instead of just matching words. The AI and Machine Learning sentences received the highest similarity score of 0.7318. I was surprised that AI can understand semantic similarity even when the exact words are different.
