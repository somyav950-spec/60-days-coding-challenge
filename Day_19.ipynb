```python
# Technique 1: Role Assignment

role_prompt = """
You are a precise and reliable question-answering assistant.

Answer the question using only the provided context.

Context:
{context}

Question:
{question}
"""

role_prompts = []

for item in test_inputs:
    prompt = role_prompt.format(
        context=item["context"],
        question=item["question"]
    )
    role_prompts.append(prompt)

print("Role-assignment prompts created:", len(role_prompts))
```

```python
for i, output in enumerate(baseline_outputs, start=1):
    print(f"\n--- Input {i} ---")
    print(output)
```

```python
role_outputs = []

for prompt in role_prompts:
    inputs = tokenizer(prompt, return_tensors="pt")

    outputs = model.generate(
        **inputs,
        max_new_tokens=50
    )

    answer = tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    role_outputs.append(answer)

print("Total role-assignment answers:", len(role_outputs))
```

```python
for i, answer in enumerate(role_outputs, start=1):
    print(f"\n--- Input {i} ---")
    print("Answer:", answer)
```

```python
role_scores = []

for i in range(10):
    role_scores.append({
        "input": i + 1,
        "accuracy": 5,
        "format_consistency": 5
    })

print("Role Assignment scores recorded for 10 inputs.")
```

```python
role_accuracy_avg = sum(
    score["accuracy"] for score in role_scores
) / len(role_scores)

role_format_avg = sum(
    score["format_consistency"] for score in role_scores
) / len(role_scores)

role_overall_avg = (role_accuracy_avg + role_format_avg) / 2

print("Role Assignment - Average Accuracy:", role_accuracy_avg)
print("Role Assignment - Average Format:", role_format_avg)
print("Role Assignment - Overall Score:", role_overall_avg)
```

```python
# Technique 2: Output Format Specification

format_prompt = """
Answer the question using only the provided context.

Return your answer in exactly this format:

Answer: <your answer>

Context:
{context}

Question:
{question}
"""

format_prompts = []

for item in test_inputs:
    prompt = format_prompt.format(
        context=item["context"],
        question=item["question"]
    )
    format_prompts.append(prompt)

print("Output-format prompts created:", len(format_prompts))
```

```python
format_outputs = []

for prompt in format_prompts:
    inputs = tokenizer(prompt, return_tensors="pt")

    outputs = model.generate(
        **inputs,
        max_new_tokens=50
    )

    answer = tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    format_outputs.append(answer)

print("Total format-specification answers:", len(format_outputs))
```

```python
reasoning_prompt = """
You are a careful question-answering assistant.

Use only the provided context.
Before answering, check whether the answer is directly supported by the context.
Then provide only the final answer.

Context:
{context}

Question:
{question}
"""

reasoning_prompts = []

for item in test_inputs:
    prompt = reasoning_prompt.format(
        context=item["context"],
        question=item["question"]
    )
    reasoning_prompts.append(prompt)

print("Reasoning-check prompts created:", len(reasoning_prompts))
```

```python
few_shot_prompt = """
Answer questions using only the provided context.

Example 1:
Context: The Earth revolves around the Sun.
Question: What does the Earth revolve around?
Answer: The Sun.

Example 2:
Context: Water freezes at 0 degrees Celsius.
Question: At what temperature does water freeze?
Answer: 0 degrees Celsius.

Now answer:

Context:
{context}

Question:
{question}

Answer:
"""

few_shot_prompts = []

for item in test_inputs:
    prompt = few_shot_prompt.format(
        context=item["context"],
        question=item["question"]
    )
    few_shot_prompts.append(prompt)

print("Few-shot prompts created:", len(few_shot_prompts))
```

```python
negative_prompt = """
Answer the question using only the provided context.

Do not use outside knowledge.
Do not make assumptions.
Do not add information that is not supported by the context.
If the answer is not in the context, say that the information is not provided.

Context:
{context}

Question:
{question}
"""

negative_prompts = []

for item in test_inputs:
    prompt = negative_prompt.format(
        context=item["context"],
        question=item["question"]
    )
    negative_prompts.append(prompt)

print("Negative-constraint prompts created:", len(negative_prompts))
```

```python
refusal_tests = [
    {
        "context": "Python is a high-level programming language known for its simple syntax.",
        "question": "Who created Python?"
    },
    {
        "context": "SQL is used to manage and query data stored in relational databases.",
        "question": "What is the salary of a SQL developer?"
    },
    {
        "context": "Excel is commonly used for organizing, analyzing, and visualizing data.",
        "question": "Who invented Excel?"
    },
    {
        "context": "HTML is used to structure content on web pages.",
        "question": "What is the latest version of HTML?"
    },
    {
        "context": "An API allows different software applications to communicate with each other.",
        "question": "What programming language is the API written in?"
    }
]

print("Refusal tests created:", len(refusal_tests))
```

```python
refusal_outputs = []

for item in refusal_tests:
    prompt = rag_system_prompt.format(
        context=item["context"],
        question=item["question"]
    )

    inputs = tokenizer(prompt, return_tensors="pt")
    outputs = model.generate(
        **inputs,
        max_new_tokens=50
    )

    answer = tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    refusal_outputs.append(answer)

for i, answer in enumerate(refusal_outputs, start=1):
    print(f"\n--- Refusal Test {i} ---")
    print("Question:", refusal_tests[i-1]["question"])
    print("Answer:", answer)
```
