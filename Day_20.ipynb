```python
import requests
import fastapi
import uvicorn
import langchain
import faiss

print("FastAPI:", fastapi.__version__)
print("LangChain:", langchain.__version__)
print("Requests:", requests.__version__)
print("All imports successful!")
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

print("Chunking pipeline ready!")
print("Chunk size:", 500)
print("Chunk overlap:", 50)
```

```python
from langchain_core.documents import Document

# Day 12 ka document
document = """
Artificial Intelligence is a field of computer science that focuses on creating
systems capable of performing tasks that normally require human intelligence.
These tasks include learning, reasoning, problem solving, understanding language,
recognizing images, making predictions, and supporting decision making.

Machine learning allows a computer system to learn patterns from data rather
than being explicitly programmed for every individual task. A machine learning
model is trained using data and can later make predictions or decisions based
on new information.

Supervised learning uses labeled data. Classification predicts categories,
while regression predicts numerical values.

Unsupervised learning works with data where predefined output labels are not
available. Clustering is a common unsupervised learning technique.

Reinforcement learning allows an agent to learn through feedback and rewards.

Deep learning is a branch of machine learning based on artificial neural
networks and is useful for image recognition, speech recognition and NLP.

Natural Language Processing enables computers to work with human language.
"""

# Day 12 chunking
recursive_chunks = text_splitter.split_text(document)

# Day 17 style metadata documents
documents = []

for i, chunk in enumerate(recursive_chunks):
    documents.append(
        Document(
            page_content=chunk,
            metadata={
                "source": "AI Research Report",
                "category": "Artificial Intelligence",
                "date": "2026-09-25",
                "document_type": "Technical",
                "chunk_id": i
            }
        )
    )

print("Total chunks:", len(documents))
print("\nFirst chunk:")
print(documents[0].page_content)

print("\nMetadata:")
print(documents[0].metadata)
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS

# Day 14: Embedding model
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

# Create FAISS vector store from our metadata documents
vectorstore = FAISS.from_documents(
    documents,
    embeddings
)

print("FAISS vector store created successfully!")
print("Total indexed documents:", len(documents))
```

```python
query = "What is machine learning?"

results = vectorstore.similarity_search(
    query,
    k=3,
    filter={
        "category": "Artificial Intelligence"
    }
)

print("Query:", query)
print("\nFiltered Results:\n")

for i, doc in enumerate(results, start=1):
    print(f"--- Result {i} ---")
    print("Source:", doc.metadata["source"])
    print("Category:", doc.metadata["category"])
    print("Chunk ID:", doc.metadata["chunk_id"])
    print("Content:", doc.page_content[:200])
    print()
```

```python
GROUNDING_PROMPT = """
You are a grounded AI assistant.

Answer the user's question ONLY using the provided context.

Rules:
1. Do not use outside knowledge.
2. If the answer is not present in the context, say:
   "I don't have enough information in the provided context."
3. Keep the answer concise and factual.
4. Mention the source when possible.

Context:
{context}

Question:
{question}

Answer:
"""

print("Grounded RAG prompt created successfully!")
```

```python
def retrieve_context(query, k=3):
    results = vectorstore.similarity_search(
        query,
        k=k,
        filter={
            "category": "Artificial Intelligence"
        }
    )

    context_parts = []

    for doc in results:
        context_parts.append(
            f"Source: {doc.metadata['source']}\n"
            f"Chunk ID: {doc.metadata['chunk_id']}\n"
            f"Content: {doc.page_content}"
        )

    return results, "\n\n".join(context_parts)


# Test retrieval
query = "What is machine learning?"

results, context = retrieve_context(query)

print("Retrieved chunks:", len(results))
print("\n" + context[:1500])
```

```python
def build_rag_prompt(query):
    results, context = retrieve_context(query)

    prompt = GROUNDING_PROMPT.format(
        context=context,
        question=query
    )

    return results, prompt


# Test
query = "What is machine learning?"

results, rag_prompt = build_rag_prompt(query)

print(rag_prompt)
```

```python
def retrieve_with_score(query, k=3):
    results = vectorstore.similarity_search_with_relevance_scores(
        query,
        k=k,
        filter={
            "category": "Artificial Intelligence"
        }
    )

    return results


query = "What is machine learning?"

results_with_scores = retrieve_with_score(query)

print("Query:", query)
print("\nRetrieved Results:\n")

for i, (doc, score) in enumerate(results_with_scores, start=1):
    print(f"--- Result {i} ---")
    print("Similarity Score:", round(score, 3))
    print("Source:", doc.metadata["source"])
    print("Chunk ID:", doc.metadata["chunk_id"])
    print("Content:", doc.page_content[:200])
    print()
```

```python
def generate_answer_with_confidence(query, threshold=0.3):
    results_with_scores = retrieve_with_score(query)

    if not results_with_scores:
        return "I don't have enough information in the provided context.", []

    best_doc, best_score = results_with_scores[0]

    if best_score < threshold:
        return (
            "Warning: Low confidence. "
            "I don't have enough information in the provided context."
        ), results_with_scores

    results, rag_prompt = build_rag_prompt(query)
    response = llm.invoke(rag_prompt)

    return response.content, results_with_scores


query = "What is machine learning?"

answer, scored_sources = generate_answer_with_confidence(query)

print("Answer:")
print(answer)

print("\nSources:")
for doc, score in scored_sources:
    print(
        f"- {doc.metadata['source']} | "
        f"Chunk {doc.metadata['chunk_id']} | "
        f"Score: {score:.3f}"
    )
```
