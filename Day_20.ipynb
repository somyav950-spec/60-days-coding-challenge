```python
from langchain_core.documents import Document
from sentence_transformers import SentenceTransformer
import numpy as np
import faiss

# Day 17 style metadata
documents = []

for i, chunk in enumerate(chunks):
    doc = Document(
        page_content=chunk,
        metadata={
            "source": "Day 12 AI Document",
            "category": "Artificial Intelligence",
            "document_type": "Technical",
            "chunk_id": i
        }
    )
    documents.append(doc)

print("Total LangChain documents:", len(documents))
print("\nFirst document metadata:")
print(documents[0].metadata)

# Day 14 embedding model
model = SentenceTransformer("all-MiniLM-L6-v2")

# Create embeddings
texts = [doc.page_content for doc in documents]

embeddings = model.encode(texts)
embeddings = np.array(embeddings, dtype="float32")

print("\nEmbedding shape:", embeddings.shape)

# Create FAISS index
dimension = embeddings.shape[1]

index = faiss.IndexFlatL2(dimension)
index.add(embeddings)

print("FAISS index created successfully!")
print("Index size:", index.ntotal)
```

```python
def semantic_search(query, top_k=3):
    # Convert query into embedding
    query_embedding = model.encode([query])
    query_embedding = np.array(query_embedding, dtype="float32")

    # Search in FAISS
    distances, indices = index.search(query_embedding, top_k)

    results = []

    for distance, idx in zip(distances[0], indices[0]):
        doc = documents[idx]

        results.append({
            "document": doc,
            "score": float(distance)
        })

    return results


# Test search
query = "What is artificial intelligence?"

results = semantic_search(query, top_k=3)

for i, result in enumerate(results, 1):
    print(f"\nResult {i}")
    print("Score:", result["score"])
    print("Text:", result["document"].page_content[:300])
    print("Metadata:", result["document"].metadata)
```

```python
def filtered_semantic_search(query, filters=None, top_k=3):
    filters = filters or {}

    # Pehle semantic search
    results = semantic_search(query, top_k=len(documents))

    filtered_results = []

    for result in results:
        doc = result["document"]
        metadata = doc.metadata

        # Category filter
        if "category" in filters:
            if metadata.get("category") != filters["category"]:
                continue

        # Document type filter
        if "document_type" in filters:
            if metadata.get("document_type") != filters["document_type"]:
                continue

        filtered_results.append(result)

    return filtered_results[:top_k]


# Test
query = "What is artificial intelligence?"

results = filtered_semantic_search(
    query,
    filters={
        "category": "Artificial Intelligence",
        "document_type": "Technical"
    },
    top_k=3
)

print("Filtered results:", len(results))

for i, result in enumerate(results, 1):
    print(f"\nResult {i}")
    print("Score:", result["score"])
    print("Text:", result["document"].page_content[:300])
    print("Metadata:", result["document"].metadata)
```

```python
def build_context(results):
    context_parts = []

    for i, result in enumerate(results, 1):
        doc = result["document"]

        context_parts.append(
            f"Source {i}:\n"
            f"{doc.page_content}\n"
            f"Metadata: {doc.metadata}\n"
        )

    return "\n".join(context_parts)


# Test
query = "What is artificial intelligence?"

results = filtered_semantic_search(
    query,
    filters={
        "category": "Artificial Intelligence",
        "document_type": "Technical"
    },
    top_k=3
)

context = build_context(results)

print("Context created successfully!")
print("\n" + context[:2000])
```

```python
def build_grounded_prompt(query, context):

    prompt = f"""
You are a helpful AI knowledge assistant.

Answer the user's question using ONLY the provided context.

If the answer is not available in the context, clearly say:
"I don't have enough information in the provided context."

Do not make up facts or use outside information.

User Question:
{query}

Retrieved Context:
{context}

Answer:
"""

    return prompt


# Test
query = "What is artificial intelligence?"

grounded_prompt = build_grounded_prompt(
    query,
    context
)

print(grounded_prompt[:3000])
```

```python
def retrieve_context(query, filters=None, top_k=3):
    results = filtered_semantic_search(
        query,
        filters=filters,
        top_k=top_k
    )

    context = build_context(results)

    return context, results


# Test
query = "What is machine learning?"

context, results = retrieve_context(
    query,
    filters={
        "category": "Artificial Intelligence",
        "document_type": "Technical"
    },
    top_k=3
)

print("RAG retrieval completed successfully!")
print("Retrieved documents:", len(results))

print("\nContext Preview:\n")
print(context[:2000])
```

```python
def get_confidence_warning(results, threshold=0.3):
    if not results:
        return "LOW CONFIDENCE: No relevant information found."

    # FAISS L2 distance mein lower value = closer match
    best_distance = results[0]["score"]

    if best_distance > threshold:
        return "LOW CONFIDENCE: The retrieved information may not be sufficiently relevant."

    return "Confidence is acceptable."


# Test
warning = get_confidence_warning(results)

print(warning)
```

```python
def prepare_rag_response(query, filters=None, top_k=3):
    # Retrieve relevant documents
    context, results = retrieve_context(
        query,
        filters=filters,
        top_k=top_k
    )

    # Confidence warning
    confidence_message = get_confidence_warning(results)

    # Source citations
    sources = []

    for i, result in enumerate(results, 1):
        doc = result["document"]

        sources.append({
            "source_id": i,
            "source": doc.metadata.get("source"),
            "category": doc.metadata.get("category"),
            "document_type": doc.metadata.get("document_type"),
            "chunk_id": doc.metadata.get("chunk_id"),
            "score": result["score"]
        })

    return {
        "query": query,
        "context": context,
        "sources": sources,
        "confidence": confidence_message
    }


# Test
response_data = prepare_rag_response(
    "What is machine learning?",
    filters={
        "category": "Artificial Intelligence",
        "document_type": "Technical"
    },
    top_k=3
)

print("RAG response prepared successfully!")

print("\nConfidence:")
print(response_data["confidence"])

print("\nSources:")
for source in response_data["sources"]:
    print(source)
```

```python
def run_rag(query, filters=None, top_k=3):
    # 1. Retrieve relevant context
    context, results = retrieve_context(
        query,
        filters=filters,
        top_k=top_k
    )

    # 2. Build grounded prompt
    prompt = build_grounded_prompt(
        query,
        context
    )

    # 3. Prepare sources
    sources = []

    for i, result in enumerate(results, 1):
        doc = result["document"]

        sources.append({
            "source_id": i,
            "source": doc.metadata.get("source"),
            "category": doc.metadata.get("category"),
            "document_type": doc.metadata.get("document_type"),
            "chunk_id": doc.metadata.get("chunk_id"),
            "score": result["score"]
        })

    # 4. Confidence information
    confidence = get_confidence_warning(results)

    return {
        "query": query,
        "prompt": prompt,
        "sources": sources,
        "confidence": confidence
    }


# Test
rag_result = run_rag(
    "What is artificial intelligence?",
    filters={
        "category": "Artificial Intelligence",
        "document_type": "Technical"
    },
    top_k=3
)

print("RAG pipeline working successfully!")

print("\nQuery:")
print(rag_result["query"])

print("\nConfidence:")
print(rag_result["confidence"])

print("\nSources:")
for source in rag_result["sources"]:
    print(source)
```

```python
def generate_answer(query, filters=None, top_k=3):
    context, results = retrieve_context(
        query,
        filters=filters,
        top_k=top_k
    )

    prompt = build_grounded_prompt(query, context)

    response = client.models.generate_content(
        model="gemini-3.8-flash",
        contents=prompt
    )

    return response.text, results


query = "What is artificial intelligence?"

answer, results = generate_answer(
    query,
    filters={
        "category": "Artificial Intelligence",
        "document_type": "Technical"
    },
    top_k=3
)

print("Answer:\n")
print(answer)
```

```python
import numpy as np

def get_confidence_warning(query, results, threshold=0.3):
    if not results:
        return "LOW CONFIDENCE: No relevant information found."

    query_embedding = model.encode([query])[0]
    query_embedding = np.array(query_embedding, dtype="float32")

    similarities = []

    for result in results:
        doc = result["document"]
        doc_embedding = model.encode([doc.page_content])[0]
        doc_embedding = np.array(doc_embedding, dtype="float32")

        cosine_similarity = np.dot(query_embedding, doc_embedding) / (
            np.linalg.norm(query_embedding) * np.linalg.norm(doc_embedding)
        )

        similarities.append(float(cosine_similarity))

    best_similarity = max(similarities)

    if best_similarity < threshold:
        return f"LOW CONFIDENCE: Similarity score is {best_similarity:.3f}"

    return f"Confidence acceptable: Similarity score is {best_similarity:.3f}"


query = "What is artificial intelligence?"

confidence = get_confidence_warning(query, results)

print(confidence)
```

```python
from fastapi import FastAPI
from pydantic import BaseModel

app = FastAPI(title="AI Knowledge Assistant")


class AskRequest(BaseModel):
    query: str


@app.post("/ask")
def ask(request: AskRequest):
    answer, results = generate_answer(
        request.query,
        filters={
            "category": "Artificial Intelligence",
            "document_type": "Technical"
        },
        top_k=3
    )

    confidence = get_confidence_warning(
        request.query,
        results
    )

    sources = []

    for i, result in enumerate(results, 1):
        doc = result["document"]

        sources.append({
            "source_id": i,
            "source": doc.metadata.get("source"),
            "category": doc.metadata.get("category"),
            "document_type": doc.metadata.get("document_type"),
            "chunk_id": doc.metadata.get("chunk_id"),
            "score": result["score"]
        })

    return {
        "query": request.query,
        "answer": answer,
        "confidence": confidence,
        "sources": sources
    }


print("FastAPI app created successfully!")
```

```python
test_queries = [
    "What is artificial intelligence?",
    "What is machine learning?",
    "What is deep learning?",
    "What is natural language processing?",
    "What is computer vision?",
    "What is neural network?",
    "What is supervised learning?",
    "What is unsupervised learning?",
    "What is reinforcement learning?",
    "What is generative AI?",
    "What are the applications of AI?",
    "How does machine learning work?",
    "What is the difference between AI and ML?",
    "What is the role of data in AI?",
    "What are the challenges of artificial intelligence?"
]

print("Total test queries:", len(test_queries))
```

```python
import requests
import pandas as pd
import time

results_15 = []

for i, query in enumerate(test_queries, 1):

    start_time = time.time()

    response = requests.post(
        "http://127.0.0.1:8000/ask",
        json={"query": query}
    )

    elapsed_time = time.time() - start_time

    results_15.append({
        "Query No": i,
        "Query": query,
        "Status Code": response.status_code,
        "Response": response.text,
        "Latency (seconds)": round(elapsed_time, 2)
    })

    print(f"Query {i}/15 completed - Status: {response.status_code}")

results_df = pd.DataFrame(results_15)

print("\n15-query testing completed!")
display(results_df)
```
