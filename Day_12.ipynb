```python
def fixed_size_chunks(text, chunk_size):
    chunks = []

    for i in range(0, len(text), chunk_size):
        chunks.append(text[i:i + chunk_size])

    return chunks
```

```python
chunks_500 = fixed_size_chunks(document, 500)

print("Number of chunks:", len(chunks_500))
print("\nFirst chunk:\n")
print(chunks_500[0])
```

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

recursive_chunks = splitter.split_text(document)

print("Number of chunks:", len(recursive_chunks))
print("\nFirst chunk:\n")
print(recursive_chunks[0])
```

```python
chunk_sizes = [100, 300, 500, 1000]
overlaps = [0, 50, 100]

results = {}

for size in chunk_sizes:
    for overlap in overlaps:

        splitter = RecursiveCharacterTextSplitter(
            chunk_size=size,
            chunk_overlap=overlap
        )

        chunks = splitter.split_text(document)

        results[(size, overlap)] = chunks

        print(
            f"Chunk Size: {size}, "
            f"Overlap: {overlap}, "
            f"Number of Chunks: {len(chunks)}"
        )
```

```python
query = "What are the main applications of artificial intelligence?"

print("Query:", query)
```

```python
def retrieve_chunks(chunks, query, top_k=3):
    query_words = set(query.lower().split())

    scored_chunks = []

    for chunk in chunks:
        chunk_words = set(chunk.lower().split())
        score = len(query_words.intersection(chunk_words))
        scored_chunks.append((score, chunk))

    scored_chunks.sort(reverse=True, key=lambda x: x[0])

    return scored_chunks[:top_k]
```

```python
for config, chunks in results.items():

    retrieved = retrieve_chunks(
        chunks,
        query,
        top_k=3
    )

    print("\n" + "=" * 60)
    print("Configuration:", config)

    for score, chunk in retrieved:
        print("\nScore:", score)
        print(chunk[:300])
```

```python
comparison = []

for config, chunks in results.items():

    retrieved = retrieve_chunks(
        chunks,
        query,
        top_k=3
    )

    best_score = retrieved[0][0] if retrieved else 0

    comparison.append({
        "Chunk Size": config[0],
        "Overlap": config[1],
        "Number of Chunks": len(chunks),
        "Best Retrieval Score": best_score
    })

comparison_df = pd.DataFrame(comparison)

comparison_df
```

```python
for i in range(len(chunks_500) - 1):

    current = chunks_500[i]
    next_chunk = chunks_500[i + 1]

    if not current.rstrip().endswith((".", "!", "?")):

        print("\n--- Boundary Example ---")
        print("Chunk:", i)

        print("End:", current[-50:])
        print("Start:", next_chunk[:50])

        print("-" * 40)

        if i >= 2:
            break
```


Final Recommendation

> Recommended Configuration: Chunk Size = 500, Chunk Overlap = 50



> The comparison shows that chunk sizes of 300 and 500 achieved a best retrieval score of 3, while the 1000-character configuration achieved a score of 4. However, the retrieval score used in this experiment is based on simple keyword matching, so a larger chunk can receive a higher score simply because it contains more words. The 500-character chunk size provides a practical balance between chunk size, number of chunks, and retrieval quality. An overlap of 50 characters helps preserve context across chunk boundaries.



Boundary Analysis

> Fixed-size chunking can split sentences or thoughts between two chunks. In this experiment, examples were observed at chunk boundaries around character positions 500, 1000, and 1500. These examples demonstrate why a small overlap can help preserve context when retrieving information.



Conclusion

> This experiment compared fixed-size and recursive character-based text chunking using different chunk sizes and overlaps. Smaller chunks produced more chunks, while larger chunks produced fewer chunks. The experiment also showed that chunk boundaries can divide meaningful sentences. Based on the comparison, 500 characters with 50-character overlap was selected as a balanced configuration for this experiment.
