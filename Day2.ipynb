```python
from collections import Counter
text = "AI is powerful and AI is used for Data Science"

words_=text.lower().split()
word_frequency= Counter(words_)
print(word_frequency)
```

```python
import string
text="Hello, AI World!,  this is python."
cleaned_text=text.lower()

cleaned_text=cleaned_text.translate(str.maketrans(" "," ",string.punctuation))
cleaned_text=" ".join(cleaned_text.split())

print(cleaned_text)
```
