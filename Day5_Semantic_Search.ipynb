```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
sentences = [
    "Data analytics helps businesses make better decisions.",
    "Python is widely used for data analysis.",
    "Excel is useful for organizing and analyzing data.",
    "SQL is used to retrieve data from databases.",
    "Data visualization makes complex information easier to understand.",
    "Power BI helps create interactive business dashboards.",
    "Pandas is a popular Python library for data manipulation.",
    "NumPy is useful for numerical calculations in Python.",
    "Data cleaning removes errors and duplicate values.",
    "Data analysts often work with large datasets.",
    "Charts and graphs help communicate data insights.",
    "A dashboard can display important business metrics.",
    "Machine learning can find patterns in data.",
    "Statistics helps analysts understand data.",
    "Businesses use data to understand customer behavior.",
    "Data analysis can help identify business trends.",
    "Good quality data improves the accuracy of analysis.",
    "SQL queries can filter and sort database records.",
    "Reports summarize important information from datasets.",
    "Data-driven decisions are based on evidence and analysis."
]

print("Total sentences:", len(sentences))
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")

print("Model loaded successfully!")
```

```python
embeddings = model.encode(sentences)

print("Embeddings created successfully!")
print("Number of vectors:", len(embeddings))
print("Vector size:", len(embeddings[0]))
```

```python
def search(query):
    query_vector = model.encode([query])
    scores = cosine_similarity(query_vector, embeddings)[0]

    top3 = scores.argsort()[-3:][::-1]

    for i in top3:
        print(sentences[i])
```

```python
search("How can businesses use data?")
```

```python
def keyword_search(query):
    results = []

    for sentence in sentences:
        count = 0

        for word in query.lower().split():
            if word in sentence.lower():
                count += 1

        results.append((count, sentence))

    results.sort(reverse=True)

    for count, sentence in results[:3]:
        print(sentence)
```

```python
keyword_search("How can businesses use data?")
```

```python
print("SEMANTIC SEARCH")
search("How can businesses use data?")

print("\nKEYWORD SEARCH")
keyword_search("How can businesses use data?")
```

## Conclusion

Semantic search understands the meaning of a query and finds related sentences.

Keyword search mainly looks for matching words.

Therefore, semantic search can give better results when different words have similar meanings.
