```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
```

```python
data = {
    "review": [
        "I loved this movie",
        "This movie was amazing",
        "The film was excellent",
        "I really enjoyed the movie",
        "What a fantastic experience",
        "The story was beautiful",
        "The actors were brilliant",
        "This was a wonderful movie",
        "I hated this movie",
        "This movie was terrible",
        "The film was boring",
        "I really disliked the movie",
        "What a horrible experience",
        "The story was disappointing",
        "The actors were bad",
        "This was a waste of time",
        "Amazing acting and great story",
        "Very enjoyable and entertaining",
        "Worst movie I have ever seen",
        "Not worth watching"
    ],

    "sentiment": [
        "positive",
        "positive",
        "positive",
        "positive",
        "positive",
        "positive",
        "positive",
        "positive",
        "negative",
        "negative",
        "negative",
        "negative",
        "negative",
        "negative",
        "negative",
        "negative",
        "positive",
        "positive",
        "negative",
        "negative"
    ]
}

df = pd.DataFrame(data)

df.head()
```

```python
X = df["review"]
y = df["sentiment"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
vectorizer = CountVectorizer()

X_train_vectorized = vectorizer.fit_transform(X_train)
X_test_vectorized = vectorizer.transform(X_test)

print("Text converted into numerical features.")
```

```python
model = MultinomialNB()

model.fit(X_train_vectorized, y_train)

print("Model training completed!")
```

```python
y_pred = model.predict(X_test_vectorized)

accuracy = accuracy_score(y_test, y_pred)

print("Model Accuracy:", accuracy)
```

```python
test_reviews = [
    "The movie was absolutely fantastic",
    "I did not enjoy this movie",
    "The acting was brilliant",
    "This was a boring and terrible film",
    "I really enjoyed the story"
]

test_vectorized = vectorizer.transform(test_reviews)

predictions = model.predict(test_vectorized)

for review, prediction in zip(test_reviews, predictions):
    print("Review:", review)
    print("Predicted Sentiment:", prediction)
    print()
```

```python
results = pd.DataFrame({
    "Review": test_reviews,
    "Predicted Sentiment": predictions
})

results
```

The model performs well on simple positive and negative sentences. However, it may fail when the review contains sarcasm, mixed emotions, spelling mistakes, or complex language.

For example, a sentence may contain both positive and negative words, which can confuse the classifier. The model also has limited understanding of context because it mainly learns word patterns from the training data.
