```python
import os
from datetime import datetime

from langchain_core.documents import Document
from langchain_openai import OpenAIEmbeddings
from langchain_community.vectorstores import FAISS

print("Libraries imported successfully!")
```

```python
documents = [
    Document(
        page_content="Artificial intelligence enables computers to perform tasks that normally require human intelligence.",
        metadata={
            "source": "AI Research Report",
            "category": "Artificial Intelligence",
            "date": "2025-06-15",
            "document_type": "Technical"
        }
    ),

    Document(
        page_content="Machine learning algorithms learn patterns from data and use those patterns to make predictions.",
        metadata={
            "source": "Machine Learning Research",
            "category": "Machine Learning",
            "date": "2026-01-20",
            "document_type": "Technical"
        }
    ),

    Document(
        page_content="Businesses use AI to improve customer engagement, marketing campaigns, and business decisions.",
        metadata={
            "source": "Business AI Guide",
            "category": "Business",
            "date": "2025-09-10",
            "document_type": "Marketing"
        }
    ),

    Document(
        page_content="AI systems can analyze customer behavior and recommend products based on user preferences.",
        metadata={
            "source": "Marketing AI Overview",
            "category": "Marketing",
            "date": "2024-08-10",
            "document_type": "Marketing"
        }
    ),

    Document(
        page_content="Recent artificial intelligence research focuses on large language models, retrieval systems, and generative AI.",
        metadata={
            "source": "Generative AI Research",
            "category": "Artificial Intelligence",
            "date": "2026-03-05",
            "document_type": "Technical"
        }
    )
]

print("Total documents:", len(documents))

for i, doc in enumerate(documents, 1):
    print(f"\nDocument {i}")
    print("Metadata:", doc.metadata)
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

print("Local embeddings created successfully!")
```

```python
vectorstore = FAISS.from_documents(
    documents,
    embeddings
)

print("FAISS vector store created successfully!")
```

```python
query = "What is artificial intelligence?"

results = vectorstore.similarity_search(query, k=3)

for i, doc in enumerate(results, 1):
    print(f"\nResult {i}")
    print("Text:", doc.page_content)
    print("Metadata:", doc.metadata)
```

```python
def filtered_retrieve(query, filters=None, k=3):
    filters = filters or {}

    # First retrieve all documents by vector similarity
    results = vectorstore.similarity_search(
        query,
        k=len(documents)
    )

    filtered_results = []

    for doc in results:
        metadata = doc.metadata

        # Category filter
        if "category" in filters:
            if metadata.get("category") != filters["category"]:
                continue

        # Document type filter
        if "document_type" in filters:
            if metadata.get("document_type") != filters["document_type"]:
                continue

        # Date filter
        if "date_after" in filters:
            doc_date = datetime.strptime(
                metadata["date"], "%Y-%m-%d"
            ).date()

            cutoff_date = datetime.strptime(
                filters["date_after"], "%Y-%m-%d"
            ).date()

            if doc_date < cutoff_date:
                continue

        filtered_results.append(doc)

    return filtered_results[:k]


print("filtered_retrieve() function created successfully!")
```

```python
query = "What is artificial intelligence?"

results = filtered_retrieve(
    query,
    filters={"category": "Artificial Intelligence"},
    k=3
)

for i, doc in enumerate(results, 1):
    print(f"\nResult {i}")
    print("Text:", doc.page_content)
    print("Metadata:", doc.metadata)
```

```python
query = "What are recent AI developments?"

results = filtered_retrieve(
    query,
    filters={
        "date_after": "2025-01-01"
    },
    k=3
)

for i, doc in enumerate(results, 1):
    print(f"\nResult {i}")
    print("Text:", doc.page_content)
    print("Metadata:", doc.metadata)
```

```python
queries = [
    {
        "query": "What is artificial intelligence?",
        "filter": {"category": "Artificial Intelligence"}
    },
    {
        "query": "How does machine learning work?",
        "filter": {"category": "Machine Learning"}
    },
    {
        "query": "How is AI used in business?",
        "filter": {"category": "Business"}
    },
    {
        "query": "What are recent AI developments?",
        "filter": {"date_after": "2025-01-01"}
    },
    {
        "query": "How can AI improve customer recommendations?",
        "filter": {"category": "Marketing"}
    }
]

for i, item in enumerate(queries, 1):

    print("=" * 70)
    print(f"QUERY {i}: {item['query']}")

    # Without metadata filter
    normal_results = vectorstore.similarity_search(
        item["query"],
        k=3
    )

    print("\nWITHOUT FILTER:")
    for j, doc in enumerate(normal_results, 1):
        print(f"{j}. {doc.metadata['source']}")

    # With metadata filter
    filtered_results = filtered_retrieve(
        item["query"],
        filters=item["filter"],
        k=3
    )

    print("\nWITH FILTER:")
    for j, doc in enumerate(filtered_results, 1):
        print(f"{j}. {doc.metadata['source']}")

    print()
```

```python
comparison = []

for i, item in enumerate(queries, 1):

    query = item["query"]
    filters = item["filter"]

    # Without filter
    normal_results = vectorstore.similarity_search(
        query,
        k=3
    )

    # With filter
    filtered_results = filtered_retrieve(
        query,
        filters=filters,
        k=3
    )

    # Number of results returned
    normal_count = len(normal_results)
    filtered_count = len(filtered_results)

    comparison.append({
        "Query": f"Query {i}",
        "Without Filter": normal_count,
        "With Filter": filtered_count,
        "Filter Used": str(filters)
    })

import pandas as pd

comparison_df = pd.DataFrame(comparison)

print("Retrieval Comparison:")
display(comparison_df)
```

```python
architecture_notes = """
DAY 17 - ADVANCED RETRIEVAL WITH METADATA

Architecture Changes from Day 15:

1. Documents now contain metadata fields:
   - source
   - category
   - date
   - document_type

2. LangChain Document objects keep the text and metadata together.

3. FAISS is used as the vector store for semantic similarity search.

4. A new filtered_retrieve() function was added.
   It supports:
   - category filtering
   - document type filtering
   - date-based filtering

5. Retrieval now works in two stages:
   Query -> Vector Similarity Search -> Metadata Filtering -> Final Results

Benefits:
Metadata filtering makes retrieval more targeted by restricting
results to the required category, document type, or date range.

Limitations / Edge Cases:

1. Missing Metadata:
   If a document does not contain required metadata such as category
   or date, the corresponding filter may not be able to use that document correctly.

2. Conflicting Categories:
   If a document belongs to multiple categories but only one category
   is stored in metadata, filtering may exclude a potentially relevant document.
"""

print(architecture_notes)
```
