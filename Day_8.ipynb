```python
import nltk
import re

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
```

```python
texts = {
    "Social Media": "OMG!!! I LOVE this new phone 😍😍 It's AMAZING!!! What do you guys think? #technology",

    "News Article": "The government announced new technology policies today. Experts said these policies will improve digital services.",

    "Customer Review": "I really loved this product!!! The quality is GOOD, but the delivery was very slow 😡. Overall, I am satisfied."
}
```

```python
for name, text in texts.items():
    print("\n==============================")
    print(name)
    print("==============================")

    sentences = sent_tokenize(text)
    words = word_tokenize(text)

    print("Sentence Tokens:")
    print(sentences)

    print("\nWord Tokens:")
    print(words)
```

```python
stop_words = set(stopwords.words('english'))

for name, text in texts.items():
    words = word_tokenize(text.lower())

    before = len(words)

    filtered_words = [
        word for word in words
        if word.isalpha() and word not in stop_words
    ]

    after = len(filtered_words)

    print("\n==============================")
    print(name)
    print("==============================")

    print("Before Stopword Removal:", words)
    print("Vocabulary/Word Count Before:", before)

    print("\nAfter Stopword Removal:", filtered_words)
    print("Vocabulary/Word Count After:", after)
```

```python
for name, text in texts.items():

    cleaned_text = re.sub(r'[^\w\s]', '', text)

    print("\n==============================")
    print(name)
    print("==============================")

    print("Before:")
    print(text)

    print("\nAfter Punctuation Removal:")
    print(cleaned_text)
```

```python
for name, text in texts.items():

    # Removing punctuation makes text cleaner and easier for NLP models
    # to process consistently during downstream tasks.
    cleaned_text = re.sub(r'[^\w\s]', '', text)

    print("\n==============================")
    print(name)
    print("==============================")

    print("Before:")
    print(text)

    print("\nAfter:")
    print(cleaned_text)
```

```python
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

for name, text in texts.items():

    words = word_tokenize(text.lower())

    words = [word for word in words if word.isalpha()]

    stemmed = [stemmer.stem(word) for word in words]
    lemmatized = [lemmatizer.lemmatize(word) for word in words]

    print("\n==============================")
    print(name)
    print("==============================")

    print("Original Words:")
    print(words)

    print("\nStemming:")
    print(stemmed)

    print("\nLemmatization:")
    print(lemmatized)
```

```python
for name, text in texts.items():

    words = word_tokenize(text.lower())
    words = [word for word in words if word.isalpha()]

    print("\n==============================")
    print(name)
    print("==============================")

    print("Word\t\tStemming\tLemmatization")
    print("-" * 50)

    for word in words:
        stem = stemmer.stem(word)
        lemma = lemmatizer.lemmatize(word)

        print(f"{word:<12}{stem:<15}{lemma}")
```

```python
for name, text in texts.items():

    print("\n\n========================================")
    print(name)
    print("========================================")

    # 1. Tokenization
    tokens = word_tokenize(text)

    # 2. Stopword removal
    stopword_removed = [
        word for word in tokens
        if word.isalpha() and word.lower() not in stop_words
    ]

    # 3. Punctuation removal
    punctuation_removed = re.sub(r'[^\w\s]', '', text)

    # 4. Stemming
    clean_words = word_tokenize(punctuation_removed.lower())
    clean_words = [word for word in clean_words if word.isalpha()]
    stemmed = [stemmer.stem(word) for word in clean_words]

    # 5. Lemmatization
    lemmatized = [lemmatizer.lemmatize(word) for word in clean_words]

    print("\n1. TOKENIZATION")
    print("Before:", text)
    print("After:", tokens)

    print("\n2. STOPWORD REMOVAL")
    print("Before:", tokens)
    print("After:", stopword_removed)

    print("\n3. PUNCTUATION REMOVAL")
    print("Before:", text)
    print("After:", punctuation_removed)

    print("\n4. STEMMING")
    print("Before:", clean_words)
    print("After:", stemmed)

    print("\n5. LEMMATIZATION")
    print("Before:", clean_words)
    print("After:", lemmatized)
```

Text Preprocessing Explanation
1. Tokenization:
Tokenization divides text into smaller units such as words and sentences. It helps NLP models process and analyse text efficiently.
2. Stopword Removal:
Stopwords are common words such as "the", "is", "and", etc. Removing them reduces unnecessary information and can reduce the vocabulary size.
3. Punctuation Removal:
Punctuation marks such as commas, full stops, and special characters are removed to make the text cleaner and more consistent for NLP processing.
4. Stemming:
Stemming reduces words to their root-like form by removing prefixes or suffixes. It is fast but may sometimes produce words that are not actual dictionary words.
5. Lemmatization:
Lemmatization converts words into their meaningful dictionary form. It is more linguistically accurate than stemming.
Conclusion
Text preprocessing converts noisy real-world text into clean and structured data. These steps help NLP models understand and process text more effectively.
